```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Original DataFrame jisme missing values hain
df_orig = df.copy()

# Sabhi columns ke missing indices ko track karna
nan_indices = {col: df_orig[df_orig[col].isna()].index for col in df_orig.columns}

# Step 0: Mean Imputation (0th Iteration)
df_curr = df_orig.copy()
for col in df_curr.columns:
    df_curr[col] = df_curr[col].fillna(df_curr[col].mean())

# Iterations run karne ka loop
max_iter = 20

for iteration in range(1, max_iter + 1):
    df_prev = df_curr.copy()
    
    # Har feature/column ke upar loop
    for col in df_curr.columns:
        # Agar iss column me originally missing values the
        if len(nan_indices[col]) > 0:
            
            # Missing row positions ko firse NaN karo
            df_curr.loc[nan_indices[col], col] = np.nan
            
            # Non-missing rows ko training ke liye aur missing rows ko prediction ke liye split karo
            train_mask = df_curr[col].notna()
            
            X_train = df_curr.loc[train_mask, [c for c in df_curr.columns if c != col]]
            y_train = df_curr.loc[train_mask, col]
            
            X_pred = df_curr.loc[~train_mask, [c for c in df_curr.columns if c != col]]
            
            # Linear Regression Model Fit karke missing values predict karo
            lr = LinearRegression()
            lr.fit(X_train, y_train)
            df_curr.loc[~train_mask, col] = lr.predict(X_pred)
            
    # Difference check karo (Convergence verify karne ke liye)
    print(f"\n--- Iteration {iteration} Difference (df_curr - df_prev) ---")
    print(df_curr - df_prev)

# Final Imputed DataFrame
print("\nFinal Imputed DataFrame:")
print(df_curr)
```
